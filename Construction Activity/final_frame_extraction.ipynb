```python
import cv2
import os

# Input folder containing video files
video_folder = "videos"
output_folder = "dataset/train/images"
os.makedirs(output_folder, exist_ok=True)

# Frame extraction counter (to avoid overwriting frames with the same name)
global_image_count = 0

# Iterate through all video files in the folder
for video_file in os.listdir(video_folder):
    if video_file.endswith((".mp4", ".avi", ".mkv", ".mov")):  # Add more extensions if needed
        video_path = os.path.join(video_folder, video_file)
        
        # Load the video
        video = cv2.VideoCapture(video_path)

        # Get the frame rate (frames per second) of the video
        fps = int(video.get(cv2.CAP_PROP_FPS))
        frame_interval = 3 * fps  # Interval for every 3 seconds

        # Frame extraction loop
        frame_count = 0

        while True:
            ret, frame = video.read()
            if not ret:  # Break if video ends
                break

            # Check if the frame is at the 3-second interval
            if frame_count % frame_interval == 0:
                image_path = os.path.join(output_folder, f"frame_{global_image_count:06}.jpg")
                cv2.imwrite(image_path, frame)  # Save frame as JPG
                print(f"Saved: {image_path}")
                global_image_count += 1

            frame_count += 1

        # Release the video capture object
        video.release()
        print(f"Frame extraction completed for {video_file}.")

print("All videos processed.")
```

```python
import os
from PIL import Image

# Folder containing the images
image_folder = "Crane"

# Find the minimum dimensions among all images
min_width, min_height = float("inf"), float("inf")

for file_name in os.listdir(image_folder):
    file_path = os.path.join(image_folder, file_name)
    if file_name.endswith((".jpg", ".jpeg", ".png", ".bmp", ".tiff")):
        with Image.open(file_path) as img:
            width, height = img.size
            min_width = min(min_width, width)
            min_height = min(min_height, height)

print(f"Minimum Dimensions: {min_width}x{min_height}")

# Ask user for confirmation
user_input = input(f"Resize all images to {min_width}x{min_height}? (Y/N): ").strip().lower()

if user_input == "y":
    # Resize all images in place
    for file_name in os.listdir(image_folder):
        file_path = os.path.join(image_folder, file_name)
        if file_name.endswith((".jpg", ".jpeg", ".png", ".bmp", ".tiff")):
            try:
                with Image.open(file_path) as img:
                    # Resize image to minimum dimensions using LANCZOS resampling
                    resized_img = img.resize((min_width, min_height), Image.Resampling.LANCZOS)
                    # Save resized image (overwrite original)
                    resized_img.save(file_path)
                    print(f"Resized in place: {file_path}")
            except Exception as e:
                print(f"Error processing {file_name}: {e}")
    print("All images resized in place successfully.")
else:
    print("Resizing aborted by user.")
```

```python
import os
import random
import string

def generate_random_filename(length=8):
    # Generate a random string of letters and digits
    characters = string.ascii_letters + string.digits  # This includes both uppercase, lowercase, and digits
    return ''.join(random.choice(characters) for _ in range(length))

def rename_images_with_random_names(directory):
    # Get a list of all files in the directory with .jpeg or .jpg extensions
    files = [f for f in os.listdir(directory) if f.endswith('.jpeg') or f.endswith('.jpg')]

    # Ensure there are files to rename
    if not files:
        print("Error: No image files found in the directory.")
        return

    # Loop through the files and rename them with random names
    for old_filename in files:
        new_filename = generate_random_filename() + '.jpeg'  # Add .jpeg extension to the random name

        old_path = os.path.join(directory, old_filename)
        new_path = os.path.join(directory, new_filename)

        # Rename the file
        os.rename(old_path, new_path)
        print(f"Renamed: {old_filename} to {new_filename}")

# Usage
directory = 'dataset/train/images'

rename_images_with_random_names(directory)
```
